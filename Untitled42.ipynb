```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from sklearn.metrics import roc_curve, classification_report, roc_auc_score
```

```python
IMG_SIZE = 224

plantvillage_dir = "/content/plantvillage/raw/color"

healthy_dir = os.path.join(plantvillage_dir, "Corn_(maize)___healthy")

diseased_dirs = [
    os.path.join(plantvillage_dir, "Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot"),
    os.path.join(plantvillage_dir, "Corn_(maize)___Common_rust_"),
    os.path.join(plantvillage_dir, "Corn_(maize)___Northern_Leaf_Blight")
]

def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        if filename.lower().endswith(('.jpg','.png','.jpeg')):
            path = os.path.join(folder, filename)
            img = load_img(path, target_size=(IMG_SIZE, IMG_SIZE))
            img = img_to_array(img) / 255.0
            images.append(img)
    return np.array(images)

def load_multiple(folders):
    all_images = []
    for folder in folders:
        all_images.extend(load_images_from_folder(folder))
    return np.array(all_images)

print("Cargando imágenes...")
x_healthy = load_images_from_folder(healthy_dir)
x_diseased = load_multiple(diseased_dirs)

print("Sanas:", x_healthy.shape)
print("Enfermas:", x_diseased.shape)
```

```python
input_img = keras.Input(shape=(224,224,3))

# -------- Encoder --------
x = layers.Conv2D(32,3,padding='same')(input_img)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(64,3,padding='same')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(128,3,padding='same')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D(2)(x)

x = layers.Conv2D(256,3,padding='same')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)
x = layers.MaxPooling2D(2)(x)

x = layers.Flatten()(x)
z = layers.Dense(64, name="latent_space")(x)

# -------- Decoder --------
x = layers.Dense(14*14*256)(z)
x = layers.Reshape((14,14,256))(x)

x = layers.UpSampling2D(2)(x)
x = layers.Conv2D(128,3,padding='same')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)

x = layers.UpSampling2D(2)(x)
x = layers.Conv2D(64,3,padding='same')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)

x = layers.UpSampling2D(2)(x)
x = layers.Conv2D(32,3,padding='same')(x)
x = layers.BatchNormalization()(x)
x = layers.ReLU()(x)

x = layers.UpSampling2D(2)(x)
output_img = layers.Conv2D(3,3,activation='sigmoid',padding='same')(x)

autoencoder = keras.Model(input_img, output_img)
```

```python
alpha = 0.8

def combined_loss(y_true, y_pred):
    mse = tf.reduce_mean(tf.square(y_true - y_pred))
    ssim = tf.reduce_mean(tf.image.ssim(y_true, y_pred, max_val=1.0))
    return alpha * mse + (1 - alpha) * (1 - ssim)

autoencoder.compile(
    optimizer=keras.optimizers.Adam(1e-4),
    loss=combined_loss
)

autoencoder.summary()
```

```python
history = autoencoder.fit(
    x_healthy,
    x_healthy,
    epochs=40,
    batch_size=16,
    validation_split=0.2
)
```

```python
recon_healthy = autoencoder.predict(x_healthy)
recon_diseased = autoencoder.predict(x_diseased)

mse_healthy = np.mean((x_healthy - recon_healthy)**2, axis=(1,2,3))
mse_diseased = np.mean((x_diseased - recon_diseased)**2, axis=(1,2,3))
```

```python
y_true = np.concatenate([
    np.zeros(len(mse_healthy)),
    np.ones(len(mse_diseased))
])

scores = np.concatenate([mse_healthy, mse_diseased])

fpr, tpr, thresholds = roc_curve(y_true, scores)

J = tpr - fpr
ix = np.argmax(J)
tau_star = thresholds[ix]

print("Umbral óptimo τ* =", tau_star)
print("TPR =", tpr[ix])
print("FPR =", fpr[ix])
print("Índice de Youden =", J[ix])
```

```python
plt.plot(fpr, tpr)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```

```python


 def predict_anomaly(image_path, model, threshold):

    # Cargar imagen
    img = load_img(image_path, target_size=(224,224))
    img_array = img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    # Reconstrucción
    reconstruction = model.predict(img_array)

    # Calcular error
    mse = np.mean((img_array - reconstruction)**2)

    # Mostrar imagen original
    plt.figure(figsize=(4,4))
    plt.imshow(img)
    plt.axis("off")

    # Mostrar resultado
    if mse > threshold:
        plt.title("Anómala")
        print("Anómala")
    else:
        plt.title("Sana")
        print("Sana")

    plt.show()
```

```python
def visualizar_ambas(x_healthy, x_diseased, model, n=3):

    # Tomar muestras aleatorias
    idx_h = np.random.choice(len(x_healthy), n)
    idx_d = np.random.choice(len(x_diseased), n)

    healthy_samples = x_healthy[idx_h]
    diseased_samples = x_diseased[idx_d]

    # Reconstrucciones
    recon_healthy = model.predict(healthy_samples)
    recon_diseased = model.predict(diseased_samples)

    plt.figure(figsize=(4*n, 8))

    # -------- SANAS --------
    for i in range(n):
        # Original sana
        plt.subplot(4, n, i + 1)
        plt.imshow(healthy_samples[i])
        plt.title("Sana - Original")
        plt.axis("off")

        # Reconstrucción sana
        plt.subplot(4, n, i + 1 + n)
        plt.imshow(recon_healthy[i])
        plt.title("Sana - Recon")
        plt.axis("off")

    # -------- ENFERMAS --------
    for i in range(n):
        # Original enferma
        plt.subplot(4, n, i + 1 + 2*n)
        plt.imshow(diseased_samples[i])
        plt.title("Enferma - Original")
        plt.axis("off")

        # Reconstrucción enferma
        plt.subplot(4, n, i + 1 + 3*n)
        plt.imshow(recon_diseased[i])
        plt.title("Enferma - Recon")
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
visualizar_ambas(x_healthy, x_diseased, autoencoder, n=3)
```

```python
def probar_imagen(image_path, model, threshold):

    # Cargar imagen
    img = load_img(image_path, target_size=(224,224))
    img_array = img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    # Reconstrucción
    reconstruction = model.predict(img_array)

    # Calcular error MSE
    mse = np.mean((img_array - reconstruction)**2)

    # Mapa de error por pixel
    error_map = np.mean((img_array - reconstruction)**2, axis=-1)[0]

    # Clasificación
    if mse > threshold:
        resultado = "Anómala"
    else:
        resultado = "Sana"

    print("Error:", mse)
    print("Umbral:", threshold)
    print("Resultado:", resultado)

    # Visualización
    plt.figure(figsize=(12,4))

    # Original
    plt.subplot(1,3,1)
    plt.imshow(img)
    plt.title("Original")
    plt.axis("off")

    # Reconstrucción
    plt.subplot(1,3,2)
    plt.imshow(reconstruction[0])
    plt.title("Reconstrucción")
    plt.axis("off")

    # Heatmap del error
    plt.subplot(1,3,3)
    plt.imshow(error_map, cmap="hot")
    plt.title("Mapa de error")
    plt.axis("off")

    plt.suptitle("Clasificación: " + resultado)
    plt.show()
```

```python

import os
import shutil
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array

source_dir = "/content/plantvillage/raw/color"
target_dir = "/content/maize_dataset"

os.makedirs(target_dir, exist_ok=True)

for folder in os.listdir(source_dir):
    if "Corn_(maize)" in folder:
        shutil.copytree(
            os.path.join(source_dir, folder),
            os.path.join(target_dir, folder),
            dirs_exist_ok=True
        )

print("Dataset de maíz listo")
```

```python
data_dir = "/content/maize_dataset"

datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_generator = datagen.flow_from_directory(
    data_dir,
    target_size=(224,224),
    batch_size=32,
    class_mode="sparse",
    subset="training"
)

val_generator = datagen.flow_from_directory(
    data_dir,
    target_size=(224,224),
    batch_size=32,
    class_mode="sparse",
    subset="validation"
)
```

```python
disease_model = keras.Sequential([

    layers.Conv2D(32,(3,3),activation='relu',input_shape=(224,224,3)),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(128,(3,3),activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(256,(3,3),activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Flatten(),

    layers.Dense(256,activation='relu'),
    layers.Dropout(0.5),

    layers.Dense(4,activation='softmax')

])

disease_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

disease_model.summary()
```

```python
history_cls = disease_model.fit(
    train_generator,
    epochs=15,
    validation_data=val_generator
)
```

```python
class_names = list(train_generator.class_indices.keys())

print(class_names)
```

```python
def diagnosticar_hoja(image_path):

    img = load_img(image_path, target_size=(224,224))
    img_array = img_to_array(img)/255.0
    img_array = np.expand_dims(img_array,axis=0)

    reconstruction = autoencoder.predict(img_array)

    mse = np.mean((img_array - reconstruction)**2)

    if mse <= tau_star:

        resultado = "Hoja sana"

    else:

        prediction = disease_model.predict(img_array)

        class_id = np.argmax(prediction)

        resultado = class_names[class_id]

    plt.imshow(img)
    plt.axis("off")
    plt.title(resultado)
    plt.show()

    print("Error MSE:", mse)
    print("Diagnóstico:", resultado)
```
